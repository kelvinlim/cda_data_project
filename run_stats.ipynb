```python
# read in the file imputed_models.csv into a df
# and print out the first 5 rows
import pandas as pd
df = pd.read_csv('imputed_models.csv')
print(df.head())
```

## Examine difference in causal graph metrics due to proportion sampling


```python
from statsmodels.formula.api import ols

import statsmodels.api as sm

# Filter the dataframe to include only rows where keepMode is 'row' and no imputation
# df_row = df[df['keepMode'] == 'row']
df_row = df[(df['keepMode'] == 'row') & (df['imputeMethod'] == 'dropna')]

# Define the model
model = ols('diceCoeff ~  C(proportion)', data=df_row).fit()

# Perform ANOVA
anova_table = sm.stats.anova_lm(model, typ=2)
print("ANOVA for dice coefficient and proportion")
print(anova_table)
```

### Post hocs for proportion


```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Perform posthoc analysis for proportion
posthoc_proportion = pairwise_tukeyhsd(df_row['diceCoeff'], df_row['proportion'])
print("Posthoc for Dice coefficient and proportion")
print(posthoc_proportion)
```

Examine differences in diceCoefficients with proportion and imputation

```python
from statsmodels.formula.api import ols

import statsmodels.api as sm

# Filter the dataframe to include only rows where keepMode is 'row'
df_row = df[df['keepMode'] == 'row']

# two way naova

# Define the model
model = ols('diceCoeff ~ C(imputeMethod) + C(proportion)', data=df_row).fit()

# Perform ANOVA
anova_table = sm.stats.anova_lm(model, typ=2)
print("ANOVA for diceCoeff, imputeMethod and proportion")
print(anova_table)
```

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Perform posthoc analysis for proportion
posthoc_proportion = pairwise_tukeyhsd(df_row['diceCoeff'], df_row['proportion'])
print("Posthoc for diceCoeff vs proportion")
print(posthoc_proportion)

# Perform posthoc analysis for imputeMethod
posthoc_imputeMethod = pairwise_tukeyhsd(df_row['diceCoeff'], df_row['imputeMethod'])
print("Posthoc diceCoeff vs imputeMethod")
print(posthoc_imputeMethod)

# Perform posthoc analysis for imputeMethod
posthoc_imputeMethod = pairwise_tukeyhsd(df_row['ESMean'], df_row['proportion'])
print("Posthoc ESMean vs proportion")
print(posthoc_imputeMethod)

# Perform posthoc analysis for imputeMethod
posthoc_imputeMethod = pairwise_tukeyhsd(df_row['ESMean'], df_row['imputeMethod'])
print("Posthoc ESMean vs imputeMethod")
print(posthoc_imputeMethod)
```

The ANOVA results indicate that while there is no difference among the imputation methods, there are significant differences in the causal models with different proportions of rows randomly removed.

## ESMean


```python
from statsmodels.formula.api import ols

import statsmodels.api as sm

# Filter the dataframe to include only rows where keepMode is 'row'
df_row = df[df['keepMode'] == 'row']

# two way naova

# Define the model
model = ols('ESMean ~ C(imputeMethod) + C(proportion) + C(imputeMethod):C(proportion)', data=df_row).fit()

# Perform ANOVA
anova_table = sm.stats.anova_lm(model, typ=2)
print("ANOVA for ESMean, imputeMethod and proportion")
print(anova_table)
```

## ESMean posthoc

```python
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Perform Tukey's HSD post-hoc test for imputeMethod
m_comp = pairwise_tukeyhsd(df_row['ESMean'], df_row['imputeMethod']) 
print("ESMean Tukey's HSD for imputeMethod:")
print(m_comp)

# Perform Tukey's HSD post-hoc test for proportion
p_comp = pairwise_tukeyhsd(df_row['ESMean'], df_row['proportion'])
print("\nESMean Tukey's HSD for proportion:")
print(p_comp)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming your data is in a pandas DataFrame called 'df' with columns 'diceCoeff', 'proportion', and 'imputeMethod'
df = df_row.copy()
# Perform the two-way ANOVA
model = ols('diceCoeff ~ C(proportion) * C(imputeMethod)', data=df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)

# Print the ANOVA table
print(anova_table)

# Plot the interaction effect
fig, ax = plt.subplots()
sm.graphics.interaction_plot(df['proportion'], df['imputeMethod'], df['diceCoeff'], ax=ax)
plt.xlabel('Proportion')
plt.ylabel('Dice Coefficient')
plt.title('Interaction Effect of Proportion and Impute Method on Dice Coefficient')
plt.show()
```
